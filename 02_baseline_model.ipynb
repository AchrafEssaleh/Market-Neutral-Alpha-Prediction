## Base line model

- it gives us reference performance that allows us to quantify whether more complex feature engineering & models actually add value
- in financ if an  advanced model does not beat the baseline  it is useless...

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
X_train = pd.read_csv("../data/raw/x_train.csv")
y_train = pd.read_csv("../data/raw/y_train.csv")

# merge for safety
df = X_train.merge(y_train, on="ID", how="inner")

print("Dataset shape:", df.shape)
```

```python
# baseline feature set 
# we only use the last 5 returns & volumes 

baseline_features = (
    [f"RET_{i}" for i in range(1, 6)] +
    [f"VOLUME_{i}" for i in range(1, 6)]
)

X = df[baseline_features]
y = df["RET"]

print("Baseline features:", baseline_features)
print("X shape:", X.shape)
print("y mean (class balance):", y.mean())
```

- let s evaluate the baseline model on unseen data to obtain an unbiased estimate of its performance...

## Train / validation split

```python
# handling missing values 
X_filled = X.fillna(0)

# TV split
X_train_bl, X_val_bl, y_train_bl, y_val_bl = train_test_split(
    X_filled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Train shape:", X_train_bl.shape)
print("Validation shape:", X_val_bl.shape)
```

```python
# Logistic Regression baseline
baseline_model = LogisticRegression(
    max_iter=1000,
    n_jobs=-1
)

baseline_model.fit(X_train_bl, y_train_bl)
```

```python
# validation predictions
y_val_pred = baseline_model.predict(X_val_bl)

# evaluation
val_accuracy = accuracy_score(y_val_bl, y_val_pred)
print("Baseline validation accuracy:", val_accuracy)
```

- Meaning our baseline beats random by  approxi 0.55 percentage points
- in finance an little improvment  can be meaningful 

This baseline model provides a reference performance using a simple and interpretable classifier with minimal features...
Any subsequent improvement obtained through feature engineering or more complex models can be directly attributed to added information rather than chance!!!
